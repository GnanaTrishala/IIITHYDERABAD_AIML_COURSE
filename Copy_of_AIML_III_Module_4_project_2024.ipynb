Creating a function for the derivative of $f(x)$ called df(x)

```python
def df(x):
    return 2 * x + 1
```

## Gradient Descent

```python
##########################
##########################

## TASK-1 : Complete the lines of code wherever marked as [REQUIRED] in this cell.

##########################
##########################

# Gradient Descent
new_x = 3
previous_x = 0
step_multiplier = 0.1
precision = 0.00001

x_list = [new_x]
slope_list = [df(new_x)]

for n in range(500):
    previous_x = new_x

    ## [REQUIRED] Compute the gradient at the current x
    gradient = df(previous_x)

    ## [REQUIRED] Update x in the opposite direction of the gradient
    new_x = previous_x - step_multiplier * gradient

    step_size = abs(new_x - previous_x)
    # print(step_size)

    x_list.append(new_x)

    ## [REQUIRED] Append the slope at the new x
    slope_list.append(df(new_x))

    if step_size < precision:
        print('Loop ran this many times:', n)
        break

print('Local minimum occurs at:', new_x)
print('Slope or df(x) value at this point is:', df(new_x))
print('f(x) value or cost at this point is:', f(new_x))
```

## Partial Derivatives & Symbolic Computation

## $$\frac{\partial f}{\partial x} = \frac{2x \ln(3)\;3^{-x^2-y^2}}{(1+3^{-x^2-y^2})^2}$$

## $$\frac{\partial f}{\partial y} = \frac{2y \ln(3)\;3^{-x^2-y^2}}{(1+3^{-x^2-y^2})^2}$$


### Partial Derivatives using Chain Rule

We have the function:

\[
f(x,y) = \frac{1}{3^{-x^2-y^2} + 1}, \quad r := 3^{-x^2-y^2}.
\]

---

**Step 1: Differentiate with respect to \(r\)**  

\[
f = \frac{1}{r+1} \quad \Rightarrow \quad \frac{df}{dr} = -\frac{1}{(r+1)^2}.
\]

---

**Step 2: Differentiate \(r\) with respect to \(x\) and \(y\)**  

Since
\[
r = 3^{-x^2-y^2} = e^{(-x^2-y^2)\ln(3)},
\]

\[
\frac{\partial r}{\partial x} = -2x \ln(3) \, r,
\qquad
\frac{\partial r}{\partial y} = -2y \ln(3) \, r.
\]

---

**Step 3: Apply Chain Rule**  

\[
\frac{\partial f}{\partial x}
= \frac{df}{dr} \cdot \frac{\partial r}{\partial x}
= \left(-\frac{1}{(r+1)^2}\right)\left(-2x \ln(3)\,r\right)
= \frac{2x \ln(3)\,r}{(r+1)^2}.
\]

\[
\frac{\partial f}{\partial y}
= \frac{df}{dr} \cdot \frac{\partial r}{\partial y}
= \left(-\frac{1}{(r+1)^2}\right)\left(-2y \ln(3)\,r\right)
= \frac{2y \ln(3)\,r}{(r+1)^2}.
\]

---

**Step 4: Substitute back \(r = 3^{-x^2-y^2}\)**  

\[
\frac{\partial f}{\partial x} = \frac{2x \ln(3)\,3^{-x^2-y^2}}{\left(1+3^{-x^2-y^2}\right)^2},
\]

\[
\frac{\partial f}{\partial y} = \frac{2y \ln(3)\,3^{-x^2-y^2}}{\left(1+3^{-x^2-y^2}\right)^2}.

```python
import numpy as np
from sympy import symbols, diff, ln

# [REQUIRED] Define the symbolic variables
a, b = symbols('a b')

# [REQUIRED] Define the function
f = lambda a, b: 1 / (1 + 3 ** (-a**2 - b**2))

# Setup
multiplier = 0.1
max_iter = 500
params = np.array([1.8, 1.0])  # initial guess

for n in range(max_iter):
    gradient_x = diff(f(a, b), a).evalf(subs={a: params[0], b: params[1]})
    gradient_y = diff(f(a, b), b).evalf(subs={a: params[0], b: params[1]})
    gradients = np.array([gradient_x, gradient_y])
    params = params - multiplier * gradients

# Results
print('Values in gradient array:', gradients)
print('Minimum occurs at x value of:', params[0])
print('Minimum occurs at y value of:', params[1])
print('The cost is:', f(params[0], params[1]))
```

```python
# Challenge: Write a python function mse(y, y_hat) that returns the MSE.
# Call the mse(y, y_hat) function and print out the MSE for the y_hat calculated above.

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# -------------------------------
# Define the MSE function
# -------------------------------
def mse(y, y_hat):
    """
    Computes Mean Squared Error (MSE) between actual values y and predicted values y_hat.
    Works for 1D or 2D arrays.
    """
    # Ensure y and y_hat are numpy arrays
    y = np.array(y)
    y_hat = np.array(y_hat)

    # Flatten arrays to 1D for safe computation
    y = y.ravel()
    y_hat = y_hat.ravel()

    # Compute Mean Squared Error
    mse_calc = np.mean((y - y_hat) ** 2)
    return mse_calc

# -------------------------------
# Example usage with linear regression
# -------------------------------

# Sample data
x_5 = np.array([[0.1, 1.2, 2.4, 3.2, 4.1, 5.7, 6.5]]).T  # shape (7,1)
y_5 = np.array([1.7, 2.4, 3.5, 3.0, 6.1, 9.4, 8.2]).reshape(7,1)  # shape (7,1)

# Create and fit Linear Regression model
regr = LinearRegression()
regr.fit(x_5, y_5)

# Predict y_hat using the trained model
y_hat = regr.predict(x_5)

# Compute MSE using the custom function
manual_mse = mse(y_5, y_hat)
print("Manually calculated MSE:", manual_mse)

# Compute MSE using sklearn for verification
sklearn_mse = mean_squared_error(y_5, y_hat)
print("MSE using sklearn:", sklearn_mse)

# -------------------------------
# Optional comparison
# -------------------------------
if np.isclose(manual_mse, sklearn_mse):
    print("\n✅ Both methods give the same MSE! Your function works perfectly.")
else:
    print("\n⚠️ The results differ — check for possible rounding or shape mismatches.")
```

## Partial Derivatives of MSE w.r.t. $\theta_0$ and $\theta_1$

## $$\frac{\partial MSE}{\partial \theta_0} = - \frac{2}{n} \sum_{i=1}^{n} \big( y^{(i)} - \theta_0 - \theta_1 x^{(i)} \big)$$

## $$\frac{\partial MSE}{\partial \theta_1} = - \frac{2}{n} \sum_{i=1}^{n} \big( y^{(i)} - \theta_0 - \theta_1 x^{(i)} \big) \big( x^{(i)} \big)$$

## MSE & Gradient Descent

```python
##########################
##########################

## TASK-10 : Complete the lines of code wherever marked as [REQUIRED] in this cell.

##########################
##########################

import numpy as np
import matplotlib.pyplot as plt

# x values, y values, array of theta parameters (theta0 at index 0 and theta1 at index 1)
def grad(x, y, thetas):
    n = y.size

    # Challenge: Create theta0_slope and theta1_slope to hold slope values from partial derivs
    theta0_slope = (-2/n) * sum(y - thetas[0] - thetas[1]*x)       # [REQUIRED]
    theta1_slope = (-2/n) * sum((y - thetas[0] - thetas[1]*x)*x)   # [REQUIRED]

    #return np.array([theta0_slope[0], theta1_slope[0]])
    #return np.append(arr=theta0_slope, values=theta1_slope)
    #return np.concatenate((theta0_slope, theta1_slope), axis=0)

    # Correct solution: return as numpy array
    return np.array([theta0_slope, theta1_slope])   # [REQUIRED]


# Gradient Descent Function
def gradient_descent(x, y, learning_rate=0.01, max_iter=1000):
    thetas = np.array([0.0, 0.0])  # initialize theta0, theta1
    for i in range(max_iter):
        gradients = grad(x, y, thetas)
        thetas = thetas - learning_rate * gradients

        # Optional: print progress every 100 iterations
        if i % 100 == 0:
            mse = np.mean((y - (thetas[0] + thetas[1]*x))**2)
            print(f"Iteration {i:4d} | θ0 = {thetas[0]:.4f}, θ1 = {thetas[1]:.4f}, MSE = {mse:.4f}")

    return thetas


# Example Data
x = np.array([0.1, 1.2, 2.4, 3.2, 4.1, 5.7, 6.5])
y = np.array([1.7, 2.4, 3.5, 3.0, 6.1, 9.4, 8.2])

# Run Gradient Descent
final_thetas = gradient_descent(x, y, learning_rate=0.01, max_iter=1000)
print("\n✅ Final optimized parameters [θ0, θ1]:", final_thetas)

# Predictions and MSE
y_pred = final_thetas[0] + final_thetas[1]*x
mse = np.mean((y - y_pred)**2)
print(f"📉 Final Mean Squared Error: {mse:.4f}")

# Plot Results
plt.scatter(x, y, color='blue', label='Actual Data')
plt.plot(x, y_pred, color='red', label='Fitted Line')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression using Gradient Descent')
plt.legend()
plt.show()
```

```python
##########################
##########################

## TASK-10 : Complete the lines of code wherever marked as [REQUIRED] in this cell.

##########################
##########################

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

# x values, y values, array of theta parameters (theta0 at index 0 and theta1 at index 1)
def grad(x, y, thetas):
    n = y.size

    # Challenge: Create theta0_slope and theta1_slope to hold slope values from partial derivs
    theta0_slope = (-2/n) * sum(y - thetas[0] - thetas[1]*x)       # [REQUIRED]
    theta1_slope = (-2/n) * sum((y - thetas[0] - thetas[1]*x)*x)   # [REQUIRED]

    #return np.array([theta0_slope[0], theta1_slope[0]])
    #return np.append(arr=theta0_slope, values=theta1_slope)
    #return np.concatenate((theta0_slope, theta1_slope), axis=0)

    # Correct solution: return as numpy array
    return np.array([theta0_slope, theta1_slope])   # [REQUIRED]


# Gradient Descent Function
def gradient_descent(x, y, learning_rate=0.01, max_iter=1000):
    thetas = np.array([0.0, 0.0])  # initialize theta0, theta1
    plot_vals = []  # to store theta0, theta1 for visualization

    for i in range(max_iter):
        gradients = grad(x, y, thetas)
        thetas = thetas - learning_rate * gradients
        plot_vals.append(thetas.copy())

        # Optional: print progress every 100 iterations
        if i % 100 == 0:
            mse = np.mean((y - (thetas[0] + thetas[1]*x))**2)
            print(f"Iteration {i:4d} | θ0 = {thetas[0]:.4f}, θ1 = {thetas[1]:.4f}, MSE = {mse:.4f}")

    return np.array(plot_vals), thetas


# Example Data
x = np.array([0.1, 1.2, 2.4, 3.2, 4.1, 5.7, 6.5])
y = np.array([1.7, 2.4, 3.5, 3.0, 6.1, 9.4, 8.2])

# Run Gradient Descent
plot_vals, final_thetas = gradient_descent(x, y, learning_rate=0.01, max_iter=1000)
print("\n✅ Final optimized parameters [θ0, θ1]:", final_thetas)

# Predictions and MSE
y_pred = final_thetas[0] + final_thetas[1]*x
mse = np.mean((y - y_pred)**2)
print(f"📉 Final Mean Squared Error: {mse:.4f}")


##########################
## 3D Surface Plot for MSE
##########################

# Create grid of theta0 and theta1 values
theta0_vals = np.linspace(-1, 5, 100)
theta1_vals = np.linspace(0, 2, 100)
plot_t0, plot_t1 = np.meshgrid(theta0_vals, theta1_vals)

# Compute cost (MSE) for each (theta0, theta1) pair
plot_cost = np.zeros_like(plot_t0)
for i in range(plot_t0.shape[0]):
    for j in range(plot_t0.shape[1]):
        t0 = plot_t0[i, j]
        t1 = plot_t1[i, j]
        y_hat = t0 + t1 * x
        plot_cost[i, j] = np.mean((y - y_hat) ** 2)

# Flatten the path of thetas from gradient descent for plotting
mse_vals = np.array([np.mean((y - (t[0] + t[1]*x))**2) for t in plot_vals])

# Plotting MSE
fig = plt.figure(figsize=[16, 12])
ax = fig.add_subplot(111, projection='3d')

ax.set_xlabel('Theta 0', fontsize=20)
ax.set_ylabel('Theta 1', fontsize=20)
ax.set_zlabel('Cost - MSE', fontsize=20)

ax.scatter(plot_vals[:, 0], plot_vals[:, 1], mse_vals, s=80, color='black')  # Path followed
ax.plot_surface(plot_t0, plot_t1, plot_cost, cmap=cm.rainbow, alpha=0.4)
plt.title("3D Visualization of MSE Surface and Gradient Descent Path", fontsize=18)
plt.show()
```
